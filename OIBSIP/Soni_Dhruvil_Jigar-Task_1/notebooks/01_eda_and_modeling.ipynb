# Iris Flower Classification
This notebook provides a compact EDA and modeling walkthrough for the production pipeline.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
project_root = Path('..').resolve()
data_path = project_root / 'data' / 'raw' / 'iris.csv'
df = pd.read_csv(data_path)
df.head()
```

```python
sns.pairplot(df.drop(columns=['Id']), hue='Species')
plt.show()
```

```python
feature_columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
target = df['Species'].str.replace('Iris-', '').str.lower()
X_train, X_test, y_train, y_test = train_test_split(df[feature_columns], target, test_size=0.2, random_state=42, stratify=target)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)
model.score(X_test_scaled, y_test)
```
